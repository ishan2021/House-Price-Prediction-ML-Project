## Phase 1: Dataset Understanding

### Objective
The objective of this phase is to understand the structure and contents of the Nepali House Price dataset before performing any preprocessing or model building. In this phase, we will inspect the dataset, identify the features, examine their data types, detect missing values, and understand the role of each attribute.

## Step 1: Import Required Libraries

We begin by importing the necessary Python libraries. Pandas is used for loading and analyzing the dataset, while NumPy provides support for numerical computations.

```python
import numpy as np 
import pandas as pd
```

## Step 2: Load the Dataset

The dataset is loaded into a Pandas DataFrame using the `read_csv()` function. A DataFrame is a tabular data structure that makes data analysis and manipulation easier.

```python
df=pd.read_csv("Nepali_house_dataset.csv")
```

## Step 3: Preview First Few Rows

We use `head()` to display the first 5 rows of the dataset. This gives an initial look at the data — the column names, sample values, and general structure of each feature.

```python
print(df.head())
```

## Step 4: Preview Last Few Rows

`tail()` shows the last 5 rows of the dataset. This is useful to verify that the data was loaded completely and to check whether the end of the file contains any anomalies or unexpected entries.

```python
print(df.tail())
```

## Step 5: Check Dataset Shape

`shape` returns a tuple `(rows, columns)` indicating the size of the dataset. Knowing the number of records and features upfront is essential for planning the preprocessing and modeling steps.

```python
print(df.shape)
```

## Step 6: List Column Names

`columns` returns all the feature names in the dataset. This helps us identify which attributes are available and verify that the column headers were read correctly.

```python
print(df.columns)
```

## Step 7: Dataset Info Overview

`df.info` provides a summary of the DataFrame, including column names, non-null counts, and data types. It is a quick way to spot missing values and verify whether each column has the expected data type.

```python
print(df.info)
```

## Step 8: Inspect Data Types

`dtypes` shows the data type of each column. Most columns here are stored as `str` (object), meaning they contain mixed or non-numeric text. Only `FLOOR`, `BEDROOM`, and `BATHROOM` are numeric (`float64`), which suggests significant cleaning will be needed for the remaining columns.

```python
print(df.dtypes)
```

## Step 9: Count Missing Values

`isnull().sum()` counts the number of missing (NaN) values in each column. This is the first step in understanding how much data is absent and which columns will require imputation or removal during the cleaning phase.

```python
print(df.isnull().sum())
```

## Step 10: Missing Value Percentage

Here we calculate the percentage of missing values per column. This gives a clearer picture of severity — for example, `PARKING` is missing ~81.5% of its values and `BUILDUP AREA` ~79%, making them candidates for either imputation or dropping during preprocessing.

```python
missing = ((df.isnull().sum() / len(df)) * 100).round(2)
print(missing)

```

## Step 11: Check for Duplicate Rows

`duplicated().sum()` counts the number of fully duplicated rows. A result of `0` here confirms there are no exact duplicate entries in the dataset, so no rows need to be removed on this basis.

```python
print(df.duplicated().sum())
```

#### Step 12: Statistical Summary of Numerical Columns

`describe()` computes summary statistics (count, mean, std, min, quartiles, max) for numeric columns only — `FLOOR`, `BEDROOM`, and `BATHROOM`. Notably, the max value of 100 for both `BEDROOM` and `BATHROOM` suggests possible outliers that will need to be addressed.

```python
print(df.describe())
```

#### Step 13: Statistical Summary of All Columns

`describe(include='all')` extends the summary to cover all columns, including string/categorical ones. For non-numeric columns it shows `count`, `unique`, `top` (most frequent value), and `freq` (its frequency). This helps spot dominant categories and how varied each field is.

```python
print(df.describe(include='all'))
```

## Step 14: Unique Location Count

`nunique()` returns the number of distinct values in the `LOCATION` column. With 514 unique locations, this column is highly granular. During modeling, it may need to be grouped or encoded carefully to avoid overfitting.

```python
# Count the unique values
a=df['LOCATION'].nunique()
print(a)
```

## Step 15: Unique Value Count for All Columns

We loop through every column and print the count of unique values. This provides a quick cardinality check — high-cardinality columns like `TITLE` (2895) and `AMENITIES` (1713) will need special treatment, while low-cardinality ones like `FACING` (36) and `FLOOR` (15) are more manageable for encoding.

```python
for col in df.columns:
    print(col)
    print(df[col].nunique())
    print()
  
```
