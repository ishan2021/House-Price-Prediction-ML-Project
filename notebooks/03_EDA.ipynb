```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns


plt.style.use("ggplot")
sns.set_theme(style="whitegrid")

%matplotlib inline
```

```python
df = pd.read_csv("../dataset/cleaned_house_price_dataset.csv")
```

## 2. Target Variable Analysis

The target variable for this project is **PRICE**. Understanding its distribution is essential because it helps identify skewness, outliers, and the overall range of house prices.

The following visualizations are used:

- Histogram to observe the distribution.
- Boxplot to identify outliers.

### Histogram of House Prices

A histogram is used to visualize the distribution of the target variable (**PRICE**). It helps identify how house prices are distributed across the dataset, whether the data is symmetric or skewed, and whether there are any unusual patterns.

#### Observation

- The distribution of house prices is **positively skewed (right-skewed)**.
- Most residential properties are priced between **1 Crore and 5 Crores NPR**.
- Only a small number of properties are priced significantly higher, resulting in a long right tail.
- The presence of these high-priced properties indicates luxury homes in the dataset.
- The distribution is not perfectly normal, suggesting that a logarithmic transformation of the target variable may be considered during model development if it improves prediction performance.

```python
plt.figure(figsize=(8,5))

sns.histplot(df["PRICE"] / 1e7, bins=40, kde=True)

plt.title("Distribution of House Prices")
plt.xlabel("Price (Crore NPR)")
plt.ylabel("Frequency")

plt.show()
```

### Boxplot of House Prices

A boxplot provides a visual summary of the distribution of house prices. It highlights the median, interquartile range (IQR), and potential outliers.

This helps identify whether the dataset contains unusually expensive properties that may influence the machine learning model.

```python

from matplotlib.ticker import FuncFormatter

def crore_formatter(x, pos):
    return f"{x/1e7:.0f} Cr"

plt.figure(figsize=(10,2))

sns.boxplot(x=df["PRICE"])

ax = plt.gca()
ax.xaxis.set_major_formatter(FuncFormatter(crore_formatter))

plt.title("Boxplot of House Prices")
plt.xlabel("Price")

plt.show()
```

```python
price_stats = (df["PRICE"] / 1e7).describe()
price_stats
```

## 3. Numerical Feature Analysis

This section explores the distribution of all numerical features in the dataset. Histograms help understand the distribution of values, while boxplots help detect skewness and potential outliers.

The numerical features analyzed are:

### 3.1 Land Area

Land area is one of the most influential factors affecting property prices. In this section, we examine the distribution of land area, identify potential outliers, and summarize its statistical characteristics.

The analysis includes:
- Histogram
- Boxplot
- Descriptive statistics

```python
plt.figure(figsize=(8,5))

sns.histplot(df["LAND AREA"], bins=30, kde=True)

plt.title("Distribution of Land Area")
plt.xlabel("Land Area (Sq. Ft.)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,2))

sns.boxplot(x=df["LAND AREA"])

plt.title("Boxplot of Land Area")
plt.xlabel("Land Area (Sq. Ft.)")

plt.show()
```

```python
df["LAND AREA"].describe()
```

#### Observation

- The distribution of land area is positively skewed.
- Most properties have land areas below approximately 5,000 sq. ft.
- A few properties have exceptionally large land areas, resulting in a long right tail.
- These observations suggest the presence of luxury or estate-sized properties in the dataset.
- The boxplot confirms the presence of several high-value outliers in land area.
- The median land area is considerably lower than the maximum value, indicating a right-skewed distribution.
- Most properties fall within a relatively small range, while a few properties have exceptionally large land areas.
- These larger land areas are likely genuine luxury or estate-sized properties rather than data entry errors.
